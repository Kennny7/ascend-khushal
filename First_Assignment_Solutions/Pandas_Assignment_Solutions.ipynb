**Q25.** Convert the hours logged column to a number.

```python
# Q25
df["hours_logged"] = pd.to_numeric(
    df["hours_logged"].astype("string").str.strip(),
    errors="coerce"
)
print(df["hours_logged"].dtype)
print(df["hours_logged"].head(10))
```

**Q26.** Convert the billing rate column to a number and remove the `$` sign first.

```python
# Q26
df["billing_rate_usd"] = pd.to_numeric(
    df["billing_rate_usd"].astype("string").str.replace("$", "", regex=False).str.strip(),
    errors="coerce"
)
print(df["billing_rate_usd"].dtype)
print(df["billing_rate_usd"].head(10))
```

**Q27.** Convert the CSAT score column to a number.

```python
# Q27
df["csat_score"] = pd.to_numeric(df["csat_score"], errors="coerce")
print(df["csat_score"].dtype)
print(df["csat_score"].head(10))
```

**Q28.** Print the average, smallest and largest value of hours logged.

```python
# Q28
print("Average:", df["hours_logged"].mean())
print("Smallest:", df["hours_logged"].min())
print("Largest:", df["hours_logged"].max())
```

**Q29.** Sort the table by hours logged from highest to lowest and show the top 10 rows.

```python
# Q29
df.sort_values("hours_logged", ascending=False).head(10)[
    ["ticket_id", "hours_logged", "status", "priority"]
]
```

**Q30.** Show the 5 tickets with the highest billable amount.

```python
# Q30
billable_clean = pd.to_numeric(
    df["billable_amount_usd"].astype("string")
      .str.replace(r"(?i)usd", "", regex=True)
      .str.replace("$", "", regex=False)
      .str.replace(",", "", regex=False)
      .str.strip(),
    errors="coerce"
)
df["billable_amount_usd"] = billable_clean
df.nlargest(5, "billable_amount_usd")[
    ["ticket_id", "billable_amount_usd", "client_name"]
]
```

**Q31.** Remove the extra spaces from the client name column.

```python
# Q31
df["client_name"] = df["client_name"].astype("string").str.strip()
print(df["client_name"].head(10).to_string(index=False))
```

**Q32.** Change the client name column to title case.

```python
# Q32
df["client_name"] = df["client_name"].str.title()
print(df["client_name"].head(10).to_string(index=False))
```

**Q33.** Convert the ticket created column to a date.

```python
# Q33
df["ticket_created_at"] = pd.to_datetime(df["ticket_created_at"], errors="coerce")
print(df["ticket_created_at"].dtype)
print(df["ticket_created_at"].head(10))
```

**Q34.** Add a new column that holds the year of the ticket.

```python
# Q34
df["ticket_year"] = df["ticket_created_at"].dt.year
print(df[["ticket_created_at", "ticket_year"]].head(10))
```

**Q35.** Add a new column that holds the month of the ticket.

```python
# Q35
df["ticket_month"] = df["ticket_created_at"].dt.month
print(df[["ticket_created_at", "ticket_month"]].head(10))
```

**Q36.** Count how many tickets were created in each year.

```python
# Q36
print(df["ticket_year"].value_counts(dropna=False).sort_index())
```

**Q37.** Add a new column called `total` equal to billing rate multiplied by hours logged.

```python
# Q37
df["total"] = df["billing_rate_usd"] * df["hours_logged"]
print(df[["billing_rate_usd", "hours_logged", "total"]].head(10))
```

**Q38.** Fill the missing values in the CSAT score column with the average CSAT score.

```python
# Q38
csat_mean = df["csat_score"].mean()
df["csat_score"] = df["csat_score"].fillna(csat_mean)
print("Average CSAT used for filling:", csat_mean)
print("Missing CSAT values remaining:", int(df["csat_score"].isna().sum()))
```

**Q39.** Drop the rows where the client name is missing.

```python
# Q39
before = len(df)
df = df.dropna(subset=["client_name"])
print("Rows before:", before)
print("Rows after:", len(df))
print("Rows removed:", before - len(df))
```

**Q45.** Draw a bar chart of ticket count by service line.

```python
# Q45
service_counts = df["serviceline"].value_counts().sort_values(ascending=False)
ax = service_counts.plot(kind="bar", figsize=(10, 5), title="Ticket Count by Service Line")
ax.set_xlabel("Service Line")
ax.set_ylabel("Ticket Count")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

**Q46.** Draw a histogram of hours logged.

**Q49.** In the SLA breached column, change Yes, Y, TRUE and 1 to True, and everything else to False.

```python
# Q49
truthy = {"yes", "y", "true", "1"}
df["sla_breached"] = (
    df["sla_breached"].astype("string").str.strip().str.lower().isin(truthy)
)
print(df["sla_breached"].value_counts(dropna=False))
```
