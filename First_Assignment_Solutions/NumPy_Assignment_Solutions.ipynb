# NumPy Assignment — IT Services Ticket Data

```python
import pandas as pd
import numpy as np

# Read the supplied data and apply the cleaning needed for the numeric matrix.
df_np = pd.read_csv("it_services_tickets_raw.csv")
df_np.columns = (
    df_np.columns
      .str.strip()
      .str.lower()
      .str.replace(r"[^a-z0-9]+", "_", regex=True)
      .str.strip("_")
)
df_np = df_np.replace(["NA", "N/A", "null", "-"], np.nan).drop_duplicates()

df_np["hours_logged"] = pd.to_numeric(df_np["hours_logged"].astype("string").str.strip(), errors="coerce")
df_np["billing_rate_usd"] = pd.to_numeric(
    df_np["billing_rate_usd"].astype("string").str.replace("$", "", regex=False).str.strip(),
    errors="coerce"
)
df_np["billable_amount_usd"] = pd.to_numeric(
    df_np["billable_amount_usd"].astype("string")
      .str.replace(r"(?i)usd", "", regex=True)
      .str.replace("$", "", regex=False)
      .str.replace(",", "", regex=False)
      .str.strip(),
    errors="coerce"
)
df_np["resolution_time_hours"] = pd.to_numeric(df_np["resolution_time_hours"], errors="coerce")
df_np["csat_score"] = pd.to_numeric(df_np["csat_score"], errors="coerce")
df_np["csat_score"] = df_np["csat_score"].fillna(df_np["csat_score"].mean())

numeric_matrix = df_np[
    ["hours_logged", "billing_rate_usd", "billable_amount_usd",
     "resolution_time_hours", "csat_score"]
].to_numpy(dtype=float)

hours = numeric_matrix[:, 0]
rates = numeric_matrix[:, 1]
billable = numeric_matrix[:, 2]
resolution_hours = numeric_matrix[:, 3]
csat = numeric_matrix[:, 4]

print("Numeric matrix shape:", numeric_matrix.shape)
print("Columns:", ["hours logged", "billing rate", "billable amount", "resolution hours", "CSAT score"])
```
